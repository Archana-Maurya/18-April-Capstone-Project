```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, classification_report

# =========================
# 📊 SAMPLE DATA (your dataset)
# =========================
data = {
    "Claim_Amount": [50000,120000,30000,70000,45000,90000,20000,150000,60000,80000],
    "Claim_Type": ["Health","Vehicle","Health","Travel","Vehicle","Health","Travel","Vehicle","Health","Travel"],
    "Location": ["Delhi","Mumbai","Delhi","Bangalore","Delhi","Mumbai","Chennai","Delhi","Bangalore","Mumbai"],
    "Previous_Claims": [2,5,1,3,0,4,1,6,2,3],
    "Fraud": ["No","Yes","No","Yes","No","Yes","No","Yes","No","Yes"]
}

df = pd.DataFrame(data)

# =========================
# 🔧 DATA PREPROCESSING
# =========================
le1 = LabelEncoder()
le2 = LabelEncoder()

df["Claim_Type"] = le1.fit_transform(df["Claim_Type"])
df["Location"] = le2.fit_transform(df["Location"])
df["Fraud"] = df["Fraud"].map({"No":0, "Yes":1})

X = df.drop("Fraud", axis=1)
y = df["Fraud"]

# =========================
# ✂ TRAIN TEST SPLIT
# =========================
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# =========================
# 🤖 MODEL TRAINING
# =========================
model = RandomForestClassifier(n_estimators=100)
model.fit(X_train, y_train)

# =========================
# 📊 PREDICTION
# =========================
y_pred = model.predict(X_test)

# =========================
# 📈 EVALUATION
# =========================
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)

print("\n==============================")
print("🤖 FRAUD DETECTION MODEL")
print("==============================")

print("\n📊 Accuracy:", round(accuracy, 2))
print("📊 Precision:", round(precision, 2))

print("\n📄 Classification Report:\n")
print(classification_report(y_test, y_pred))

# =========================
# 🔍 FEATURE IMPORTANCE
# =========================
print("\n⭐ Feature Importance:")
for i, col in enumerate(X.columns):
    print(col, ":", model.feature_importances_[i])
```
